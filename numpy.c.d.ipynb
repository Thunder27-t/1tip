```python
from time import time
import numpy as np

def measure_time(func):
  def wrapper(*args, **kwargs):
    start_time = time()
    result = func(*args, **kwargs)
    end_time = time()
    exec_time = end_time - start_time
    print(f'Funkcja {func.__name__} wykonała się w {exec_time:.5f}s')
    return exec_time
  return wrapper
```

```python
size = 1_000_000
py_list = list(range(size))
np_array = np.arange(size)
```

```python
@measure_time
def py_mul(py_list):
  py_list= [x * 2 for x in py_list ]

@measure_time
def np_mul(np_array):
  np_result = np_array * 2

time1 = py_mul(py_list)
time2 = np_mul(np_array)
print(f'Wykorzystywanie np przyspiesza obliczenia {time1/time2:.2f}x')
```

```python
@measure_time
def standard_sum(py_list):
  py_sum = sum(py_list)
@measure_time
def numpy_sum(np_array):
  np_sum = np.sum(np_array)

  time1 = standard_sum(py_list)
  time2 = numpy_sum(np_array)
  print(f'Stosunek czsów obliczeń : {time1/time2:.2f}x')
```
